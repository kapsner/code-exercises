```r
#| echo: false
#| include: false
# Daten laden (Setup aus ML-Einführung)
req_pkg <- c(
  "pak",
  "riskCommunicator",
  "data.table",
  "tinyplot",
  "see",
  "fastshap",
  "splitTools",
  "tidymodels",
  "kernlab",
  "parsnip",
  "parallel",
  "yardstick",
  "rpart.plot",
  "ranger",
  "DALEXtra",
  "shapviz",
  "vip",
  "bonsai",
  "lightgbm"
)
for (r in req_pkg) {
  if (!(r %in% installed.packages()[, "Package"])) {
    if (r %in% c("vip", "fastshap")) {
      # install from r-universe
      install.packages(
        r,
        repos = c(
          "https://bgreenwell.r-universe.dev",
          "https://cloud.r-project.org"
        )
      )
    } else {
      install.packages(r)
    }
  }
}

dataset_full <- riskCommunicator::framingham |>
  data.table::data.table() # Daten einlesen
# Subset: Basisuntersuchung
dataset <- dataset_full[get("PERIOD") == 1, ]

# Relevante Spalten definieren
use_cols <- c(
  "SEX",
  "TOTCHOL",
  "AGE",
  "SYSBP",
  "CURSMOKE",
  "CIGPDAY",
  "BMI",
  "DIABETES",
  "HYPERTEN"
)
# Relevante Spalten filtern, fehlende Werte entfernen
dataset <- dataset[,
  .SD,
  .SDcols = use_cols
] |>
  na.omit()

# Transformieren der katgeorialen Variablen
# "SEX" "CURSMOKE" "CIGPDAY" "DIABETES" "HYPERTEN"
cat_vars <- use_cols[c(1, 5, 8, 9)]
# Datentyp "factor" ändern
dataset[,
  (cat_vars) := lapply(
    X = .SD,
    FUN = factor
  ),
  .SDcols = cat_vars
]
rm(dataset_full, cat_vars, use_cols, r, req_pkg)

# Die Funktion erzeugt eine Liste mit Indices für die jeweiligen Datensets
# Das stratifizierte Splitten anhand der Zielvariable "HYPERTEN" soll deren
# gleichmäßige Verteilung in den Teildatensätzen sicherstellen.
data_splits <- splitTools::partition(
  y = dataset$HYPERTEN,
  p = c(train = 0.7, validation = 0.15, test = 0.15),
  type = "stratified",
  seed = 123
)

# Teildatensätze für Regressions- und Klassifizierungs-Beispiele
# Regression: Zielvariable "SYSBP" --> Entfernen von "HYPERTEN"
dataset_reg <- dataset[,
  .SD,
  .SDcols = setdiff(colnames(dataset), "HYPERTEN")
]
# Klassifizierung: Zielvariable "HYPERTEN" --> Entfernen von "SYSBP"
dataset_cls <- dataset[,
  .SD,
  .SDcols = setdiff(colnames(dataset), "SYSBP")
]
```

```r
#| echo: true
# Definition des SVM-Modells
svm_spec <- parsnip::svm_rbf() |>
  parsnip::set_mode("classification") |>
  parsnip::set_engine("kernlab", scaled = TRUE)

# Fitten des Modells auf die Daten
# "cost" entspricht dem Tuning-Parameter 'C'
set.seed(123)
svm_m1 <- svm_spec |>
  parsnip::set_args( # bei Fehler: "C-bsvc" anstatt "C-svc"
    cost = 500, prob.model = TRUE, type = "C-svc") |>
  parsnip::fit(HYPERTEN ~ .,
  data = dataset_cls[data_splits$train, ])
```

```r
#| echo: true
svm_m1
```

```r
#| echo: true
# Vorhersagen werden auf dem Test-Datensatz berechnet
svm_m1_preds <- svm_m1 |> parsnip::augment(
  new_data = dataset_cls[data_splits$test, ])

# Accuracy und AU-ROC
acc <- svm_m1_preds |> yardstick::accuracy(
  truth = HYPERTEN, estimate = .pred_class)
roc <- svm_m1_preds |> yardstick::roc_auc(
  truth = HYPERTEN, .pred_1, event_level = "second")

# ROC-Kurve
p <- svm_m1_preds |> yardstick::roc_curve(
  truth = HYPERTEN, .pred_1, event_level = "second")
rbind(acc, roc)
```

```r
#| echo: true
#| out-width: 540px
#| fig-align: center
ggplot2::autoplot(p)
```

```r
#| echo: true
# Definition des SVM-Modells
svm_spec <- parsnip::svm_rbf() |>
  parsnip::set_mode("regression") |>
  parsnip::set_engine("kernlab", scaled = TRUE)

# Fitten des Modells auf die Daten
# "cost" entspricht dem Tuning-Parameter 'C'
set.seed(123)
svm_m2 <- svm_spec |>
  parsnip::set_args(cost = 500) |>
  parsnip::fit(SYSBP ~ ., data = dataset_reg[data_splits$train, ])

# Vorhersagen werden auf dem Test-Datensatz berechnet
svm_m2_preds <- svm_m2 |>
  parsnip::augment(new_data = dataset_reg[data_splits$test, ])
# RMSE
rmse <- svm_m2_preds |>
  yardstick::rmse(truth = SYSBP, estimate = .pred)
```

```r
#| echo: true
svm_m2

# RMSE
rmse
```

```r
#| echo: true
# Definition des Regression Trees
tree_spec <- parsnip::decision_tree() |>
  parsnip::set_engine("rpart")
reg_tree_spec <- tree_spec |>
  parsnip::set_mode("regression")

# Fitten des Modells auf die Daten
# Zielvariable: 'SYSBP' aufgrund des Regressions-Settings
set.seed(123)
tree_m1 <- reg_tree_spec |>
  parsnip::set_args(model = TRUE) |>
  parsnip::fit(
    SYSBP ~ .,
    data = dataset_reg[data_splits$train, ]
  )

# Vorhersagen werden auf dem Test-Datensatz berechnet
tree_m1_preds <- tree_m1 |>
  parsnip::augment(
    new_data = dataset_reg[data_splits$test, ]
  )
```

```r
#| echo: true
tree_m1
```

```r
#| echo: true
#| out-width: 650px
#| fig-align: center
tree_m1 |>
  parsnip::extract_fit_engine() |> rpart.plot::rpart.plot()
```

```r
#| echo: true
# RMSE
tree_m1_preds |>
  yardstick::rmse(
    truth = SYSBP,
    estimate = .pred
  )
```

```r
#| echo: true
# Definition des Classification Trees
cls_tree_spec <- tree_spec |> parsnip::set_mode("classification")

cntrl <- rpart::rpart.control(
  cp = 0.0075, # Schwellenwert für Informationszuwachs
  minbucket = 20) # Mindestanzahl an Beobachtungen im Endknoten

# Fitten des Modells auf die Daten
# Zielvariable: 'HYPERTEN' aufgrund des Klassifizierungs-Settings
set.seed(123)
tree_m2 <- cls_tree_spec |>
  parsnip::set_args(model = TRUE, control = cntrl) |>
  parsnip::fit(HYPERTEN ~ ., data = dataset_cls[data_splits$train, ])

# Vorhersagen werden auf dem Test-Datensatz berechnet
tree_m2_preds <- tree_m2 |>
  parsnip::augment(new_data = dataset_cls[data_splits$test, ])

acc <- tree_m2_preds |> # Accuracy und AU-ROC
  yardstick::accuracy(truth = HYPERTEN, estimate = .pred_class)
roc <- tree_m2_preds |>
  yardstick::roc_auc(truth = HYPERTEN, .pred_1, event_level = "second")

p <- tree_m2_preds |> # ROC-Kurve
  yardstick::roc_curve(truth = HYPERTEN, .pred_1, event_level = "second")
```

```r
#| echo: true
tree_m2
rbind(acc, roc)
```

```r
#| echo: true
#| out-width: 600px
#| fig-align: center
tree_m2 |>
  parsnip::extract_fit_engine() |>
  rpart.plot::rpart.plot()
```

```r
#| echo: true
#| out-width: 550px
#| fig-align: center
ggplot2::autoplot(p)
```

```r
#| echo: false
#| include: false
# Parallelisieren, wenn möglich
nc <- parallel::detectCores()
# Verwende Hälfte der verfügbaren CPU threads,
# jedoch mindestens 2
ncores <- pmax(2, ceiling(nc) / 2)
```

```r
#| echo: true
# Definition des Random Forests
rf_spec <- parsnip::rand_forest(
  trees = 1000, # B
  mtry = floor(sqrt(ncol(dataset_cls))) # m
) |> parsnip::set_engine("ranger")

cls_rf_spec <- rf_spec |>
  parsnip::set_args(
    replace = FALSE,
    num.threads = ncores
  ) |>
  parsnip::set_mode("classification")

# Fitten des Modells auf die Daten
# Zielvariable: 'HYPERTEN' aufgrund des
# Klassifizierungs-Settings
set.seed(123)
rf_m1 <- cls_rf_spec |>
  parsnip::fit(
    HYPERTEN ~ .,
    data = dataset_cls[data_splits$train, ])
```

```r
#| echo: true
# Vorhersagen werden auf dem Test-Datensatz berechnet
rf_m1_preds <- rf_m1 |>
  parsnip::augment(
    new_data = dataset_cls[data_splits$test, ])

# Accuracy und AU-ROC
acc <- rf_m1_preds |>
  yardstick::accuracy(
    truth = HYPERTEN, estimate = .pred_class)
roc <- rf_m1_preds |>
  yardstick::roc_auc(
    truth = HYPERTEN, .pred_1, event_level = "second")

# ROC-Kurve
p <- rf_m1_preds |>
  yardstick::roc_curve(
    truth = HYPERTEN, .pred_1, event_level = "second")
```

```r
#| echo: true
rf_m1
rbind(acc, roc)
```

```r
#| echo: true
#| out-width: 550px
#| fig-align: center
ggplot2::autoplot(p)
```

```r
#| echo: true
# RMSE
rf_m2_preds |>
  yardstick::rmse(
    truth = SYSBP,
    estimate = .pred
  )
```

```r
#| echo: true
library(bonsai)
# Definition des lightgbm Modells
lgb_spec <- parsnip::boost_tree(
  trees = 1000, # B
  mtry = floor(sqrt(ncol(dataset_cls))), # m
  loss_reduction = 0.05
) |> parsnip::set_engine("lightgbm")
cls_lgb_spec <- lgb_spec |>
  parsnip::set_args(
    early_stopping_round = 50,
    num_threads = ncores
  ) |>
  parsnip::set_mode("classification")

# Fitten des Modells auf die Daten
# Zielvariable: 'HYPERTEN' (Klassifizierungs-Setting)
set.seed(123)
lgb_m1 <- cls_lgb_spec |>
  parsnip::fit(
    HYPERTEN ~ .,
    data = dataset_cls[data_splits$train, ])
```

```r
# Vorhersagen werden auf dem Test-Datensatz berechnet
lgb_m1_preds <- lgb_m1 |>
  parsnip::augment(
    new_data = dataset_cls[data_splits$test, ])

# Accuracy und AU-ROC
acc <- lgb_m1_preds |>
  yardstick::accuracy(
    truth = HYPERTEN, estimate = .pred_class)
roc <- lgb_m1_preds |>
  yardstick::roc_auc(
    truth = HYPERTEN, .pred_1,
    event_level = "second")

# ROC-Curve
p <- lgb_m1_preds |>
  yardstick::roc_curve(
    truth = HYPERTEN,
    .pred_1, event_level = "second")
```

```r
#| echo: true
# Funktion für Vorhersagen mit
# Random Forest (ranger)
pfun <- function(object, newdata) {
  # muss numerischen Vektor zurückgeben
  as.vector(
    predict(object, newdata)
  )$predictions
}

X_mat <- subset(
  x = dataset_reg[data_splits$train, ],
  select = -SYSBP)
baseline_val <- mean(pfun(
    object = parsnip::extract_fit_engine(rf_m2),
    newdata = X_mat))
```

```r
#| echo: true
# Berechnung der Shapley-Werte
# für Random Forest
s <- fastshap::explain(
  object = parsnip::extract_fit_engine(rf_m2),
  X = X_mat,
  newdata = subset(
    x = dataset_reg[data_splits$test, ],
    select = -SYSBP),
  pred_wrapper = pfun,
  nsim = 10,
  baseline = baseline_val
)

# Erzeugen des Visualisierungs-Objects
shap_freq <- shapviz::shapviz(
  object = s,
  X = dataset_reg[data_splits$test, ],
  baseline = baseline_val
)
```

```r
#| echo: true
#| out-width: 750px
#| fig-align: center
shapviz::sv_importance(shap_freq)
```
